```python
# Preprocesamiento de atípicos con IQR: capeo por columna numérica

numeric_df = df.select_dtypes(include=[np.number]).copy()
df_preprocesado = df.copy()
outlier_summary = []

for col in numeric_df.columns:
    q1 = numeric_df[col].quantile(0.25)
    q3 = numeric_df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (df_preprocesado[col] < lower_bound) | (df_preprocesado[col] > upper_bound)
    outlier_count = outlier_mask.sum()

    df_preprocesado[col] = df_preprocesado[col].clip(lower=lower_bound, upper=upper_bound)

    outlier_summary.append({
        "column": col,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": int(outlier_count)
    })

outliers_df = pd.DataFrame(outlier_summary).sort_values("outlier_count", ascending=False)
print("Resumen de atípicos antes del capeo:")
print(outliers_df)

print("\nForma original:", df.shape)
print("Forma procesada:", df_preprocesado.shape)
print("\nVista previa del dataset preprocesado:")
df_preprocesado.head()
```

```python
# Técnica 2: Valores negativos a cero en columnas de stock

print("\n=== TÉCNICA 2: Tratar Valores Negativos ===")
stock_cols = ['stock_actual', 'stock_minimo', 'stock_cierre', 'stock_proyectado_7d', 'stock_proyectado_14d']
consumo_cols = [col for col in df_preprocesado.columns if 'consumo' in col.lower() or 'demanda' in col.lower()]

print(f"Columnas de stock: {stock_cols}")
print(f"Columnas de consumo/demanda: {consumo_cols}")

# Reemplazar negativos por 0 en stocks
for col in stock_cols:
    if col in df_preprocesado.columns:
        negativos = (df_preprocesado[col] < 0).sum()
        if negativos > 0:
            print(f"  {col}: {negativos} valores negativos → 0")
            df_preprocesado[col] = df_preprocesado[col].clip(lower=0)

# Reemplazar negativos por 0 en consumo/demanda
for col in consumo_cols:
    if col in df_preprocesado.columns:
        negativos = (df_preprocesado[col] < 0).sum()
        if negativos > 0:
            print(f"  {col}: {negativos} valores negativos → 0")
            df_preprocesado[col] = df_preprocesado[col].clip(lower=0)

print("✓ Valores negativos tratados")
```

```python
# Técnica 5: Feature Engineering (creación de variables derivadas)

print("\n=== TÉCNICA 5: Feature Engineering ===")

df_features = df_preprocesado.copy()

# Ratio stock actual / stock mínimo
df_features['ratio_stock_minimo'] = (df_features['stock_actual'] / (df_features['stock_minimo'] + 1)).fillna(0)

# Cobertura: stock actual / consumo diario (cuántos días de stock hay)
df_features['dias_cobertura'] = (df_features['stock_actual'] / (df_features['consumo_diario'] + 1)).fillna(0)

# Diferencia entre ingresos y salidas
df_features['neto_movimiento'] = df_features['ingresos_stock'] - df_features['salidas_stock']

# Variabilidad del consumo (consumo promedio 7d vs consumo diario)
df_features['variabilidad_consumo'] = (abs(df_features['consumo_diario'] - df_features['consumo_promedio_7d']) / (df_features['consumo_promedio_7d'] + 1)).fillna(0)

# Porcentaje de ocupación del albergue como proxy de demanda
df_features['es_ocupacion_alta'] = (df_features['porcentaje_ocupacion'] > df_features['porcentaje_ocupacion'].median()).astype(int)

# Volumen de pacientes con alta prioridad
df_features['ratio_pacientes_alta_prioridad'] = (df_features['pacientes_prioridad_alta'] / (df_features['pacientes_total'] + 1)).fillna(0)

print("Variables derivadas creadas:")
print(f"  - ratio_stock_minimo: stock actual / stock mínimo")
print(f"  - dias_cobertura: días de stock disponible")
print(f"  - neto_movimiento: ingresos - salidas")
print(f"  - variabilidad_consumo: variabilidad del consumo")
print(f"  - es_ocupacion_alta: flag de ocupación alta")
print(f"  - ratio_pacientes_alta_prioridad: proporción de pacientes prioritarios")

print(f"\nForma con nuevas features: {df_features.shape}")
print("\nEsta adísticas de las nuevas variables:")
print(df_features[['ratio_stock_minimo', 'dias_cobertura', 'variabilidad_consumo']].describe())
```
